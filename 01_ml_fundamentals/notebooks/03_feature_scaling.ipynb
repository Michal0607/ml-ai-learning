```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
```

```python
X = np.array([[1, 100], [2, 200], [3, 300], [4, 400], [5, 500]], dtype = float)
```

```python
print(X.mean(axis = 0))
print(X.std(axis = 0))
```

```python
mean = X.mean(axis = 0)
std = X.std(axis = 0)

X_scaled = (X - mean) / std

print(X_scaled)
print(X_scaled.mean(axis = 0))
print(X_scaled.std(axis = 0))
```

```python
X = np.array([[1, 120], [2, 190], [3, 330], [4, 390], [5, 520]], dtype = float)

mean = X.mean(axis = 0)
std = X.std(axis = 0)

X_scaled = (X - mean) / std

print(X_scaled)
print(X_scaled.mean(axis = 0))
print(X_scaled.std(axis = 0))
```

```python
fig, axes = plt.subplots(1, 2, figsize = (12, 5))

axes[0].set_title("Original features")
axes[0].scatter(X[:, 0], X[:, 1])
axes[0].set_xlabel("1 feature")
axes[0].set_ylabel("2 feature")
axes[1].set_title("Standarized features")
axes[1].set_xlabel("1 feature")
axes[1].set_ylabel("2 feature")
axes[1].scatter(X_scaled[:, 0], X_scaled[:, 1])
plt.tight_layout()
plt.show()
```
